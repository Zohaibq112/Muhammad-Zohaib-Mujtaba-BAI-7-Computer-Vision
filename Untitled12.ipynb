```python
import os

ROOT_DIR = "/kaggle/input/skin-cancer9-classesisic/Skin cancer ISIC The International Skin Imaging Collaboration"

print("Found inside ROOT_DIR:", os.listdir(ROOT_DIR))
```

```python
import os
from dataclasses import dataclass, field
from typing import List
import torch


@dataclass
class RunConfig:
    train_dir: str = os.path.join(ROOT_DIR, "Train")
    test_dir: str = os.path.join(ROOT_DIR, "Test")
    out_dir: str = "./results"

    # keep only these 5 lesion types (edit freely)
    keep_classes: List[str] = field(default_factory=lambda: [
        "nevus",
        "melanoma",
        "basal cell carcinoma",
        "pigmented benign keratosis",
        "squamous cell carcinoma",
    ])

    val_share: float = 0.15
    seed: int = 42

    img_size: int = 224
    batch_size: int = 32
    warmup_epochs: int = 2      # head-only training
    finetune_epochs: int = 3    # full-network training  (2 + 3 = 5 total)
    lr: float = 1e-4
    workers: int = 2

    backbones: List[str] = field(default_factory=lambda: [
        "alexnet", "vgg16", "vgg19", "resnet18",
        "resnet50", "resnet101", "densenet121", "efficientnet_b0",
    ])


CFG = RunConfig()
os.makedirs(CFG.out_dir, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Train classes on disk:", os.listdir(CFG.train_dir))
print("Test classes on disk :", os.listdir(CFG.test_dir))
print("Device:", DEVICE)
print("Class subset in use:", CFG.keep_classes)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset, DataLoader
from torchvision.datasets.folder import default_loader
import torchvision.transforms as T

VALID_SUFFIXES = (".jpg", ".jpeg", ".png")


def index_lesion_folders(base_dir, wanted_classes):
    """Return (list of (path, label) pairs, sorted class names, name->idx map)
    scanning only `wanted_classes` under `base_dir`."""
    sorted_classes = sorted(wanted_classes)
    name_to_idx = {cname: i for i, cname in enumerate(sorted_classes)}

    pairs = []
    for cname in sorted_classes:
        folder = os.path.join(base_dir, cname)
        if not os.path.isdir(folder):
            raise FileNotFoundError(f"Missing expected folder: {folder}")
        for fname in sorted(os.listdir(folder)):
            if fname.lower().endswith(VALID_SUFFIXES):
                pairs.append((os.path.join(folder, fname), name_to_idx[cname]))
    return pairs, sorted_classes, name_to_idx


def build_transform(is_training):
    norm = T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    if is_training:
        return T.Compose([
            T.Resize((CFG.img_size, CFG.img_size)),
            T.RandomHorizontalFlip(0.5),
            T.RandomVerticalFlip(0.5),
            T.RandomAffine(degrees=0, translate=(0.2, 0.2), shear=0.2),
            T.RandomResizedCrop(CFG.img_size, scale=(0.8, 1.0)),
            T.ToTensor(),
            norm,
        ])
    return T.Compose([T.Resize((CFG.img_size, CFG.img_size)), T.ToTensor(), norm])


class LesionSubsetDataset(Dataset):
    """Thin wrapper around a pre-scanned list of (path, label) pairs."""

    def __init__(self, pairs, transform):
        self.pairs = pairs
        self.transform = transform

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, i):
        path, label = self.pairs[i]
        img = default_loader(path)
        return self.transform(img), label


def make_splits():
    all_train_pairs, class_names, _ = index_lesion_folders(CFG.train_dir, CFG.keep_classes)
    test_pairs, _, _ = index_lesion_folders(CFG.test_dir, CFG.keep_classes)

    y = np.array([lbl for _, lbl in all_train_pairs])
    idx = np.arange(len(all_train_pairs))
    tr_idx, va_idx = train_test_split(
        idx, test_size=CFG.val_share, stratify=y, random_state=CFG.seed
    )

    tr_pairs = [all_train_pairs[i] for i in tr_idx]
    va_pairs = [all_train_pairs[i] for i in va_idx]

    tr_set = LesionSubsetDataset(tr_pairs, build_transform(True))
    va_set = LesionSubsetDataset(va_pairs, build_transform(False))
    te_set = LesionSubsetDataset(test_pairs, build_transform(False))
    return tr_set, va_set, te_set, class_names


train_set, val_set, test_set, LESION_CLASSES = make_splits()
print("Class order (== label indices):", LESION_CLASSES)
print(f"train={len(train_set)}  val={len(val_set)}  test={len(test_set)}")

train_dl = DataLoader(train_set, batch_size=CFG.batch_size, shuffle=True, num_workers=CFG.workers)
val_dl = DataLoader(val_set, batch_size=CFG.batch_size, shuffle=False, num_workers=CFG.workers)
test_dl = DataLoader(test_set, batch_size=CFG.batch_size, shuffle=False, num_workers=CFG.workers)
```

```python
import torch.nn as nn
from torchvision import models

MODEL_BUILDERS = {
    "alexnet": lambda: models.alexnet(weights="IMAGENET1K_V1"),
    "vgg16": lambda: models.vgg16(weights="IMAGENET1K_V1"),
    "vgg19": lambda: models.vgg19(weights="IMAGENET1K_V1"),
    "resnet18": lambda: models.resnet18(weights="IMAGENET1K_V1"),
    "resnet50": lambda: models.resnet50(weights="IMAGENET1K_V2"),
    "resnet101": lambda: models.resnet101(weights="IMAGENET1K_V2"),
    "densenet121": lambda: models.densenet121(weights="IMAGENET1K_V1"),
    "efficientnet_b0": lambda: models.efficientnet_b0(weights="IMAGENET1K_V1"),
}


def replace_classifier_head(net, arch_name, num_classes):
    """Swap the final layer of `net` for a fresh nn.Linear(..., num_classes)."""
    if arch_name == "alexnet" or arch_name.startswith("vgg"):
        net.classifier[6] = nn.Linear(net.classifier[6].in_features, num_classes)
    elif arch_name.startswith("resnet"):
        net.fc = nn.Linear(net.fc.in_features, num_classes)
    elif arch_name == "densenet121":
        net.classifier = nn.Linear(net.classifier.in_features, num_classes)
    elif arch_name == "efficientnet_b0":
        net.classifier[1] = nn.Linear(net.classifier[1].in_features, num_classes)
    else:
        raise ValueError(f"No head-swap rule for {arch_name}")
    return net


def strip_classifier_head(net, arch_name):
    """Replace the final layer with nn.Identity() so forward() returns pooled features."""
    if arch_name == "alexnet" or arch_name.startswith("vgg"):
        net.classifier[6] = nn.Identity()
    elif arch_name.startswith("resnet"):
        net.fc = nn.Identity()
    elif arch_name == "densenet121":
        net.classifier = nn.Identity()
    elif arch_name == "efficientnet_b0":
        net.classifier[1] = nn.Identity()
    return net


def new_model(arch_name, num_classes):
    net = MODEL_BUILDERS[arch_name]()
    return replace_classifier_head(net, arch_name, num_classes)


def freeze_backbone(net, freeze: bool):
    """freeze=True -> only the last child module stays trainable."""
    for p in net.parameters():
        p.requires_grad = not freeze
    last_block = list(net.children())[-1]
    for p in last_block.parameters():
        p.requires_grad = True
```

```python
import copy
import time
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score


def score_predictions(y_true, y_pred, y_proba=None):
    out = {
        "Accuracy (%)": round(accuracy_score(y_true, y_pred) * 100, 2),
        "Precision (%)": round(precision_score(y_true, y_pred, average="weighted", zero_division=0) * 100, 2),
        "Recall (%)": round(recall_score(y_true, y_pred, average="weighted", zero_division=0) * 100, 2),
        "F1-Score (%)": round(f1_score(y_true, y_pred, average="weighted", zero_division=0) * 100, 2),
    }
    if y_proba is not None:
        try:
            out["AUC (%)"] = round(roc_auc_score(y_true, y_proba, multi_class="ovr", average="weighted") * 100, 2)
        except ValueError:
            out["AUC (%)"] = float("nan")
    return out


class Trainer:
    def __init__(self, net, device):
        self.net = net.to(device)
        self.device = device
        self.criterion = nn.CrossEntropyLoss()

    def _pass_over(self, loader, optimizer=None):
        training = optimizer is not None
        self.net.train() if training else self.net.eval()
        loss_sum, correct, seen = 0.0, 0, 0
        with torch.set_grad_enabled(training):
            for x, y in loader:
                x, y = x.to(self.device), y.to(self.device)
                logits = self.net(x)
                loss = self.criterion(logits, y)
                if training:
                    optimizer.zero_grad()
                    loss.backward()
                    optimizer.step()
                loss_sum += loss.item() * x.size(0)
                correct += (logits.argmax(1) == y).sum().item()
                seen += x.size(0)
        return loss_sum / seen, correct / seen

    def fit(self, train_loader, val_loader, warmup_epochs, finetune_epochs, lr, tag=""):
        # Stage 1: head only
        freeze_backbone(self.net, freeze=True)
        opt = torch.optim.Adam(filter(lambda p: p.requires_grad, self.net.parameters()), lr=lr)
        for e in range(warmup_epochs):
            tr_loss, tr_acc = self._pass_over(train_loader, opt)
            va_loss, va_acc = self._pass_over(val_loader, None)
            print(f"  [{tag} warmup {e+1}/{warmup_epochs}] train_acc={tr_acc:.3f} val_acc={va_acc:.3f}")

        # Stage 2: everything unfrozen
        freeze_backbone(self.net, freeze=False)
        opt = torch.optim.Adam(self.net.parameters(), lr=lr)
        sched = torch.optim.lr_scheduler.ReduceLROnPlateau(
            opt, mode="min", factor=0.1 ** 0.5, patience=7, min_lr=0.5e-6
        )
        best_acc, best_state = 0.0, copy.deepcopy(self.net.state_dict())
        for e in range(finetune_epochs):
            tr_loss, tr_acc = self._pass_over(train_loader, opt)
            va_loss, va_acc = self._pass_over(val_loader, None)
            sched.step(va_loss)
            if va_acc > best_acc:
                best_acc, best_state = va_acc, copy.deepcopy(self.net.state_dict())
            print(f"  [{tag} finetune {e+1}/{finetune_epochs}] train_acc={tr_acc:.3f} val_acc={va_acc:.3f}")

        self.net.load_state_dict(best_state)
        return self.net

    @torch.no_grad()
    def predict_probs(self, loader):
        self.net.eval()
        y_true, y_prob = [], []
        for x, y in loader:
            probs = torch.softmax(self.net(x.to(self.device)), dim=1).cpu().numpy()
            y_true.append(y.numpy())
            y_prob.append(probs)
        return np.concatenate(y_true), np.concatenate(y_prob)

    def test_metrics(self, loader):
        y_true, y_prob = self.predict_probs(loader)
        y_pred = y_prob.argmax(axis=1)
        return score_predictions(y_true, y_pred, y_prob)
```

```python
table1_rows = []

for arch in CFG.backbones:
    print(f"\n=== {arch} ===")
    t0 = time.time()

    net = new_model(arch, len(LESION_CLASSES))
    trainer = Trainer(net, DEVICE)
    trainer.fit(train_dl, val_dl, CFG.warmup_epochs, CFG.finetune_epochs, CFG.lr, tag=arch)

    row = trainer.test_metrics(test_dl)
    row["Model"] = arch
    row["Train Time (min)"] = round((time.time() - t0) / 60, 1)
    table1_rows.append(row)

    torch.save(trainer.net.state_dict(), os.path.join(CFG.out_dir, f"{arch}_best.pt"))
    pd.DataFrame(table1_rows).to_csv(os.path.join(CFG.out_dir, "table1_transfer_learning_results.csv"), index=False)

table1_df = pd.DataFrame(table1_rows)
table1_df
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC, LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from xgboost import XGBClassifier

BEST_ARCH = table1_df.sort_values("Accuracy (%)", ascending=False).iloc[0]["Model"]
print("Feature extractor chosen for Table 2:", BEST_ARCH)

feature_net = new_model(BEST_ARCH, len(LESION_CLASSES))
feature_net.load_state_dict(torch.load(os.path.join(CFG.out_dir, f"{BEST_ARCH}_best.pt"), map_location=DEVICE))
feature_net = strip_classifier_head(feature_net, BEST_ARCH).to(DEVICE).eval()


@torch.no_grad()
def pull_features(loader, net):
    feats, labels = [], []
    for x, y in loader:
        feats.append(net(x.to(DEVICE)).cpu().numpy())
        labels.append(y.numpy())
    return np.concatenate(feats), np.concatenate(labels)


train_dl_ordered = DataLoader(train_set, batch_size=CFG.batch_size, shuffle=False, num_workers=CFG.workers)
print("Pulling deep features for the training split...")
Xtr, ytr = pull_features(train_dl_ordered, feature_net)
print("Pulling deep features for the test split...")
Xte, yte = pull_features(test_dl, feature_net)
print("Feature dimensionality:", Xtr.shape[1])

CLASSICAL_MODELS = [
    ("Logistic Regression", LogisticRegression(max_iter=2000, n_jobs=-1)),
    ("Decision Tree", DecisionTreeClassifier(random_state=CFG.seed)),
    ("Random Forest", RandomForestClassifier(n_estimators=300, random_state=CFG.seed, n_jobs=-1)),
    ("K-Nearest Neighbors (KNN)", KNeighborsClassifier(n_neighbors=5)),
    ("Linear SVM", CalibratedClassifierCV(LinearSVC(max_iter=5000))),
    ("RBF-SVM", SVC(kernel="rbf", probability=True, random_state=CFG.seed)),
    ("XGBoost", XGBClassifier(n_estimators=300, eval_metric="mlogloss", random_state=CFG.seed)),
]

table2_rows = []
for clf_name, estimator in CLASSICAL_MODELS:
    print(f"\n=== {clf_name} on deep features ===")
    estimator.fit(Xtr, ytr)
    preds = estimator.predict(Xte)
    try:
        proba = estimator.predict_proba(Xte)
    except AttributeError:
        proba = None
    row = score_predictions(yte, preds, proba)
    row["Feature Extractor"] = "Deep Features"
    row["Classifier"] = clf_name
    table2_rows.append(row)
    pd.DataFrame(table2_rows).to_csv(os.path.join(CFG.out_dir, "table2_classifier_results.csv"), index=False)

table2_df = pd.DataFrame(table2_rows)
table2_df
```

```python
from thop import profile

TIMING_REPEATS = 50


def checkpoint_size_mb(path):
    return os.path.getsize(path) / (1024 ** 2)


def average_latency_ms(net, image_size, repeats=TIMING_REPEATS, warmup=10):
    net.eval()
    dummy = torch.randn(1, 3, image_size, image_size, device=DEVICE)
    with torch.no_grad():
        for _ in range(warmup):
            net(dummy)
    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    t0 = time.time()
    with torch.no_grad():
        for _ in range(repeats):
            net(dummy)
    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    return (time.time() - t0) / repeats * 1000


table3_rows = []
acc_lookup = table1_df.set_index("Model")["Accuracy (%)"]

for arch in CFG.backbones:
    print(f"\n=== benchmarking {arch} ===")
    ckpt_path = os.path.join(CFG.out_dir, f"{arch}_best.pt")

    net = new_model(arch, len(LESION_CLASSES)).to(DEVICE)
    net.load_state_dict(torch.load(ckpt_path, map_location=DEVICE))

    dummy = torch.randn(1, 3, CFG.img_size, CFG.img_size, device=DEVICE)
    flops, n_params = profile(net, inputs=(dummy,), verbose=False)
    latency_ms = average_latency_ms(net, CFG.img_size)

    table3_rows.append({
        "Model": arch,
        "Parameters (M)": round(n_params / 1e6, 2),
        "Model Size (MB)": round(checkpoint_size_mb(ckpt_path), 2),
        "FLOPs (G)": round(flops / 1e9, 2),
        "Inference Time (ms)": round(latency_ms, 2),
        "Accuracy (%)": acc_lookup.get(arch, None),
    })
    pd.DataFrame(table3_rows).to_csv(os.path.join(CFG.out_dir, "table3_efficiency_results.csv"), index=False)

table3_df = pd.DataFrame(table3_rows)
table3_df
```
