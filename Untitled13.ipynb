```python
import os, cv2, json, random, itertools, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.applications import (
    ResNet50, DenseNet121, EfficientNetB0, MobileNetV2, VGG16, InceptionV3
)
from tensorflow.keras.applications.resnet50 import preprocess_input as pp_resnet50
from tensorflow.keras.applications.densenet import preprocess_input as pp_densenet121
from tensorflow.keras.applications.efficientnet import preprocess_input as pp_efficientnetb0
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as pp_mobilenetv2
from tensorflow.keras.applications.vgg16 import preprocess_input as pp_vgg16
from tensorflow.keras.applications.inception_v3 import preprocess_input as pp_inceptionv3

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support, f1_score,
    balanced_accuracy_score, roc_auc_score, confusion_matrix, classification_report
)
from sklearn.preprocessing import label_binarize

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

RESULTS_DIR = Path("results")
FIG_DIR = RESULTS_DIR / "figures"
HIST_DIR = RESULTS_DIR / "history"
for d in [RESULTS_DIR, FIG_DIR, HIST_DIR]:
    d.mkdir(parents=True, exist_ok=True)

print("TensorFlow:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))
```

```python
# Locate metadata CSV and image folders (kagglehub layout may nest one extra folder level)
def find_file(root, name):
    for p in Path(root).rglob(name):
        return p
    return None

metadata_path = find_file(DATA_PATH, "HAM10000_metadata.csv")
assert metadata_path is not None, "Could not find HAM10000_metadata.csv - check DATA_PATH"
df_meta = pd.read_csv(metadata_path)

# Build a dict mapping image_id -> full file path (images are split across two folders)
image_paths = {}
for img_path in Path(DATA_PATH).rglob("*.jpg"):
    image_paths[img_path.stem] = str(img_path)

df_meta["path"] = df_meta["image_id"].map(image_paths)
df_meta = df_meta.dropna(subset=["path"]).reset_index(drop=True)

print("Total labeled images found:", len(df_meta))
df_meta.head()
```

```python
# HAM10000 diagnostic classes:
# akiec: Actinic keratoses | bcc: Basal cell carcinoma | bkl: Benign keratosis
# df: Dermatofibroma | mel: Melanoma | nv: Melanocytic nevi | vasc: Vascular lesions

# ---- STUDENT CONFIG: choose the classes / subset size to use ----
CLASSES_TO_USE = ["akiec", "bcc", "bkl", "df", "mel", "nv", "vasc"]  # use all 7, or narrow e.g. ["mel","nv","bkl"]
IMAGES_PER_CLASS = 200   # cap per class for manageable runtime; set None to use all available images
# -------------------------------------------------------------

df_sub = df_meta[df_meta["dx"].isin(CLASSES_TO_USE)].copy()

if IMAGES_PER_CLASS is not None:
    df_sub = (df_sub.groupby("dx", group_keys=False)
                     .apply(lambda x: x.sample(min(len(x), IMAGES_PER_CLASS), random_state=SEED)))

df_sub = df_sub.reset_index(drop=True)
CLASS_NAMES = sorted(df_sub["dx"].unique())
NUM_CLASSES = len(CLASS_NAMES)
class_to_idx = {c: i for i, c in enumerate(CLASS_NAMES)}
df_sub["label"] = df_sub["dx"].map(class_to_idx)

print(f"Selected classes: {CLASS_NAMES}")
print(f"Total images in working subset: {len(df_sub)}")
df_sub["dx"].value_counts()
```

```python
# Fixed dataset split - reused identically across ALL experiments (fair comparison requirement)
train_df, temp_df = train_test_split(df_sub, test_size=0.30, stratify=df_sub["label"], random_state=SEED)
val_df, test_df   = train_test_split(temp_df, test_size=0.50, stratify=temp_df["label"], random_state=SEED)

print("Train:", len(train_df), " Val:", len(val_df), " Test:", len(test_df))
train_df.to_csv(RESULTS_DIR / "train_split.csv", index=False)
val_df.to_csv(RESULTS_DIR / "val_split.csv", index=False)
test_df.to_csv(RESULTS_DIR / "test_split.csv", index=False)
```

```python
IMG_SIZE = 224  # spatial size fed to the CNN backbones

def load_rgb(path, size=IMG_SIZE):
    img = cv2.imread(path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)
    return img

def filter_none(img):
    return img

def filter_average(img, k=5):
    return cv2.blur(img, (k, k))

def filter_gaussian(img, k=5, sigma=0):
    return cv2.GaussianBlur(img, (k, k), sigma)

def filter_median(img, k=5):
    return cv2.medianBlur(img, k)

def filter_sharpen(img):
    kernel = np.array([[0, -1, 0],
                        [-1, 5, -1],
                        [0, -1, 0]])
    return cv2.filter2D(img, -1, kernel)

def filter_sobel(img):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    mag = cv2.magnitude(sx, sy)
    mag = np.uint8(255 * mag / (mag.max() + 1e-8))
    return cv2.cvtColor(mag, cv2.COLOR_GRAY2RGB)  # 3-channel so it still fits pretrained input shape

FILTERS = {
    "No Filter":  filter_none,
    "Average":    filter_average,
    "Gaussian":   filter_gaussian,
    "Median":     filter_median,
    "Sharpening": filter_sharpen,
    "Sobel":      filter_sobel,
}
```

```python
MODEL_REGISTRY = {
    "resnet50":       (ResNet50,       pp_resnet50),
    "densenet121":    (DenseNet121,    pp_densenet121),
    "efficientnetb0": (EfficientNetB0, pp_efficientnetb0),
    "mobilenetv2":    (MobileNetV2,    pp_mobilenetv2),
    "vgg16":          (VGG16,          pp_vgg16),
    "inceptionv3":    (InceptionV3,    pp_inceptionv3),
}

# ---- STUDENT CONFIG: the 3 best models identified in Lab Activity 1 ----
BEST_MODELS = ["resnet50", "densenet121", "efficientnetb0"]  # <-- replace with your Lab 1 findings
# -------------------------------------------------------------------------

def build_model(model_key, num_classes=NUM_CLASSES, img_size=IMG_SIZE, freeze_base=True):
    base_cls, _ = MODEL_REGISTRY[model_key]
    base = base_cls(include_top=False, weights="imagenet", input_shape=(img_size, img_size, 3))
    base.trainable = not freeze_base

    inputs = layers.Input(shape=(img_size, img_size, 3))
    x = base(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = models.Model(inputs, outputs)
    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model
```

```python
def make_dataset(dataframe, filter_fn, preprocess_fn, batch_size=32, shuffle=False, augment=False):
    paths = dataframe["path"].values
    labels = dataframe["label"].values

    def _load(path, label):
        path = path.numpy().decode("utf-8")
        img = load_rgb(path)
        img = filter_fn(img)
        img = img.astype(np.float32)
        img = preprocess_fn(img)
        return img, label

    def _tf_load(path, label):
        img, label = tf.py_function(_load, [path, label], [tf.float32, tf.int64])
        img.set_shape((IMG_SIZE, IMG_SIZE, 3))
        label.set_shape([])
        return img, label

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if shuffle:
        ds = ds.shuffle(buffer_size=len(dataframe), seed=SEED)
    ds = ds.map(_tf_load, num_parallel_calls=tf.data.AUTOTUNE)

    if augment:
        def _aug(img, label):
            img = tf.image.random_flip_left_right(img)
            img = tf.image.random_flip_up_down(img)
            return img, label
        ds = ds.map(_aug, num_parallel_calls=tf.data.AUTOTUNE)

    ds = ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)
    return ds
```

```python
BATCH_SIZE = 32
EPOCHS = 3  # quick run - lowered from 15 for faster iteration

def train_one_experiment(model_key, filter_name):
    tf.keras.backend.clear_session()
    _, preprocess_fn = MODEL_REGISTRY[model_key]
    filter_fn = FILTERS[filter_name]

    train_ds = make_dataset(train_df, filter_fn, preprocess_fn, BATCH_SIZE, shuffle=True, augment=True)
    val_ds   = make_dataset(val_df,   filter_fn, preprocess_fn, BATCH_SIZE, shuffle=False)
    test_ds  = make_dataset(test_df,  filter_fn, preprocess_fn, BATCH_SIZE, shuffle=False)

    model = build_model(model_key)

    callbacks = [
        tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=1, min_lr=1e-6),
    ]

    t0 = time.time()
    history = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS,
                         callbacks=callbacks, verbose=1)
    train_time = time.time() - t0

    tag = f"{model_key}_{filter_name.replace(' ', '_')}"
    with open(HIST_DIR / f"{tag}.json", "w") as f:
        json.dump(history.history, f)

    return model, history, test_ds, tag, train_time
```

```python
def evaluate_model(model, test_ds, tag):
    y_true, y_prob = [], []
    for batch_x, batch_y in test_ds:
        probs = model.predict(batch_x, verbose=0)
        y_prob.append(probs)
        y_true.append(batch_y.numpy())
    y_true = np.concatenate(y_true)
    y_prob = np.concatenate(y_prob)
    y_pred = np.argmax(y_prob, axis=1)

    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)
    precision, recall, f1, support = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0)
    macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except ValueError:
        auc = np.nan

    per_class = classification_report(y_true, y_pred, target_names=CLASS_NAMES,
                                       zero_division=0, output_dict=True)

    cm = confusion_matrix(y_true, y_pred)

    metrics = {
        "accuracy": acc, "precision": precision, "recall": recall,
        "f1": f1, "macro_f1": macro_f1, "balanced_accuracy": bal_acc, "auc": auc
    }

    with open(RESULTS_DIR / f"{tag}_per_class.json", "w") as f:
        json.dump(per_class, f, indent=2)
    np.save(RESULTS_DIR / f"{tag}_confusion_matrix.npy", cm)

    return metrics, cm, per_class
```

```python
def plot_training_curves(history, tag):
    hist = history.history
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(hist["accuracy"], label="Train")
    axes[0].plot(hist["val_accuracy"], label="Val")
    axes[0].set_title(f"{tag} - Accuracy"); axes[0].set_xlabel("Epoch"); axes[0].legend()

    axes[1].plot(hist["loss"], label="Train")
    axes[1].plot(hist["val_loss"], label="Val")
    axes[1].set_title(f"{tag} - Loss"); axes[1].set_xlabel("Epoch"); axes[1].legend()

    plt.tight_layout()
    plt.savefig(FIG_DIR / f"{tag}_curves.png", dpi=150)
    plt.close(fig)

def plot_confusion_matrix(cm, tag):
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
    plt.title(f"Confusion Matrix - {tag}")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.tight_layout()
    plt.savefig(FIG_DIR / f"{tag}_confmat.png", dpi=150)
    plt.close()
```

```python
all_results = []

for model_key in BEST_MODELS:
    for filter_name in FILTERS.keys():
        print("=" * 70)
        print(f"Running: model={model_key} | filter={filter_name}")
        print("=" * 70)

        model, history, test_ds, tag, train_time = train_one_experiment(model_key, filter_name)
        metrics, cm, per_class = evaluate_model(model, test_ds, tag)

        plot_training_curves(history, tag)
        plot_confusion_matrix(cm, tag)

        row = {
            "Model": model_key,
            "Filter": filter_name,
            "Accuracy": metrics["accuracy"],
            "Precision": metrics["precision"],
            "Recall": metrics["recall"],
            "F1-score": metrics["f1"],
            "Macro-F1": metrics["macro_f1"],
            "Balanced_Accuracy": metrics["balanced_accuracy"],
            "AUC": metrics["auc"],
            "Train_time_sec": round(train_time, 1),
        }
        all_results.append(row)

        # free memory between runs
        del model
        tf.keras.backend.clear_session()

results_df = pd.DataFrame(all_results)
results_df.to_csv(RESULTS_DIR / "comparison_table.csv", index=False)
results_df
```
